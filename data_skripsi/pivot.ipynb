# Data Jumlah Puskesmas #

```python
import pandas as pd

# ==============================================================================
# DATA
# ==============================================================================
df_puskesmas = pd.read_csv('data_jumlah_puskesmas.csv')

print("=========== DATA ===========")
df_puskesmas
```

```python
# Pivot Rawat Inap
df_rawat_inap = df_puskesmas.pivot_table(
    index= ['kode_kabupaten_kota', 'nama_kabupaten_kota'],
    columns='tahun',
    values='PELAYANAN_RAWAT_INAP',
    aggfunc='sum'
).reset_index()

df_rawat_inap.columns.name = None
df_rawat_inap.head(10)
```

```python
# Pivot Non Rawat Inap
df_non_rawat_inap = df_puskesmas.pivot_table(
    index= ['kode_kabupaten_kota', 'nama_kabupaten_kota'],
    columns='tahun',
    values='PELAYANAN_NON_RAWAT_INAP',
    aggfunc='sum'
).reset_index()

df_non_rawat_inap.columns.name = None
df_non_rawat_inap.head(10)
```

# Data Jumlah Penduduk

```python
import pandas as pd

# ==============================================================================
# DATA
# ==============================================================================
df_penduduk = pd.read_csv('data_jumlah_penduduk.csv')

print("=========== DATA ===========")
df_penduduk
```

```python
# Pivot Penduduk
df_penduduk = df_penduduk.pivot_table(
    index='nama_kabupaten_kota',
    columns='tahun',
    values='JUMLAH_PENDUDUK',
    aggfunc='sum'
).reset_index()

df_penduduk.columns.name = None
df_penduduk.head(10)
```

# Simpan Data satu Excel

```python
import pandas as pd

# Nama file Excel
nama_file = 'dataset_pivot.xlsx'

with pd.ExcelWriter(nama_file, engine='openpyxl') as writer:

    # Sheet 1
    df_rawat_inap.to_excel(
        writer,
        sheet_name='Rawat_Inap',
        index=False
    )

    # Sheet 2
    df_non_rawat_inap.to_excel(
        writer,
        sheet_name='Non_Rawat_Inap',
        index=False
    )

    # Sheet 3
    df_penyakitmenular.to_excel(
        writer,
        sheet_name='Penyakit_Menular',
        index=False
    )

    # Sheet 4
    df_penyakittidakmenular.to_excel(
        writer,
        sheet_name='Penyakit_Tidak_Menular',
        index=False
    )

    # Sheet 5
    df_tenagakesehatan_bidan.to_excel(
        writer,
        sheet_name='Tenaga_Kesehatan_Bidan',
        index=False
    )

    # Sheet 6
    df_tenagakesehatan_dokter.to_excel(
        writer,
        sheet_name='Tenaga_Kesehatan_Dokter',
        index=False
    )

# Sheet 7
    df_tenagakesehatan_perawat.to_excel(
        writer,
        sheet_name='Tenaga_Kesehatan_Perawat',
        index=False
    )

# Sheet 8
    df_penduduk.to_excel(
        writer,
        sheet_name='Penduduk',
        index=False
    )
print(f"File berhasil disimpan sebagai: {nama_file}")
```
