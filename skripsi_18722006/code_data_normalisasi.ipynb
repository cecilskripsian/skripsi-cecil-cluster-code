# NORMALISASI ROBUST #

```python
import pandas as pd
from sklearn.preprocessing import RobustScaler

# ==============================================================================
# DATA
# ==============================================================================
df_normalisasi = pd.read_csv('dataset_final_skripsi.csv')
kolom_string = [
    'kode_provinsi',
    'kode_kabupaten_kota',
]

df_normalisasi[kolom_string] = df_normalisasi[kolom_string].astype(str)

kolom_numerik = [
    col for col in df_normalisasi.select_dtypes(include='number').columns
    if col != 'tahun'
]

print("=========== DATA ===========")
df_normalisasi
```

```python
# ==============================================================================
# NORMALISASI ROBUST
# ==============================================================================
scaler = RobustScaler()

df_normalisasi[kolom_numerik] = scaler.fit_transform(df_normalisasi[kolom_numerik])

print("========== HASIL NORMALISASI ROBUST ==========")
df_normalisasi
```

```python
# =====================================================
# SIMPAN DATA
# =====================================================
df_normalisasi.to_csv(
    "dataset_final_skripsi_normalisasi.csv",
    index=False
)
```
