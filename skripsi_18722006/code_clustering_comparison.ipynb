# KOMPARASI K-MEANS & AGGLOMERATIVE HIERARCHICAL CLUSTERING

```python
import pandas as pd

# ==============================================================================
# DATA K-MEANS
# ==============================================================================
df_kmeans = pd.read_csv('data_clustering_kmeans_normalized.csv')
kolom_string = [
    'kode_provinsi',
    'kode_kabupaten_kota',
]

df_kmeans[kolom_string] = df_kmeans[kolom_string].astype(str)

kolom_numerik = [
    col for col in df_kmeans.select_dtypes(include='number').columns
    if col != 'tahun'
]

print("=========== DATA ===========")
df_kmeans
```

```python
import pandas as pd

# ==============================================================================
# DATA AGGLOMERATIVE HIERARCHICAL CLUSTERING (AHC)
# ==============================================================================
df_ahc = pd.read_csv('data_clustering_agglomerative_normalized.csv')
kolom_string = [
    'kode_provinsi',
    'kode_kabupaten_kota',
]

df_ahc[kolom_string] = df_ahc[kolom_string].astype(str)

kolom_numerik = [
    col for col in df_ahc.select_dtypes(include='number').columns
    if col != 'tahun'
]

print("=========== DATA ===========")
df_ahc
```

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

# ==========================================================
# PERBANDINGAN EVALUASI K-MEANS DAN AHC (K = 2)
# ==========================================================

# K-MEANS
silhouette_kmeans = silhouette_score(
    df_kmeans[kolom_numerik],
    df_kmeans['cluster']
)

dbi_kmeans = davies_bouldin_score(
    df_kmeans[kolom_numerik],
    df_kmeans['cluster']
)

chi_kmeans = calinski_harabasz_score(
    df_kmeans[kolom_numerik],
    df_kmeans['cluster']
)

# AHC
silhouette_ahc = silhouette_score(
    df_ahc[kolom_numerik],
    df_ahc['cluster']
)

dbi_ahc = davies_bouldin_score(
    df_ahc[kolom_numerik],
    df_ahc['cluster']
)

chi_ahc = calinski_harabasz_score(
    df_ahc[kolom_numerik],
    df_ahc['cluster']
)


# ==========================================================
# TABEL PERBANDINGAN
# ==========================================================

df_perbandingan = pd.DataFrame({
    'Metode': ['K-Means', 'AHC'],
    'Silhouette Score': [silhouette_kmeans, silhouette_ahc],
    'Davies-Bouldin Index': [dbi_kmeans, dbi_ahc],
    'Calinski-Harabasz Index': [chi_kmeans, chi_ahc]
})

print("========== PERBANDINGAN EVALUASI K=2 ==========")
df_perbandingan.round(4)
```

```python
# ==========================================================
# PERBANDINGAN 
# ==========================================================

print("\n========== PERBANDINGAN ==========")

# Silhouette → semakin tinggi semakin baik
if silhouette_kmeans > silhouette_ahc:
    print("Silhouette Score : K-Means lebih tinggi")
elif silhouette_ahc > silhouette_kmeans:
    print("Silhouette Score : AHC lebih tinggi")
else:
    print("Silhouette Score : Nilai sama")


# DBI → semakin rendah semakin baik
if dbi_kmeans < dbi_ahc:
    print("Davies-Bouldin Index : K-Means lebih rendah")
elif dbi_ahc < dbi_kmeans:
    print("Davies-Bouldin Index : AHC lebih rendah")
else:
    print("Davies-Bouldin Index : Nilai sama")


# CH → semakin tinggi semakin baik
if chi_kmeans > chi_ahc:
    print("Calinski-Harabasz Index : K-Means lebih tinggi")
elif chi_ahc > chi_kmeans:
    print("Calinski-Harabasz Index : AHC lebih tinggi")
else:
    print("Calinski-Harabasz Index : Nilai sama")
```
