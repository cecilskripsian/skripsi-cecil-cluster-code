```python
import pandas as pd

# ==============================================================================
# DATA
# ==============================================================================
df_final = pd.read_csv('data_clustering_kmeans.csv')
kolom_string = [
    'kode_provinsi',
    'kode_kabupaten_kota',
]

df_final[kolom_string] = df_final[kolom_string].astype(str)

kolom_numerik = [
    col for col in df_final.select_dtypes(include='number').columns
    if col != 'tahun'
]

print("=========== DATA ===========")
df_final
```

```python
df_final['nama_kabupaten_kota'] = df_final['nama_kabupaten_kota'].str.replace(
    'KABUPATEN ', '', regex=False
)

df_final['nama_kabupaten_kota'] = (
    df_final['nama_kabupaten_kota']
    .str.replace('KOTA SURABAYA', 'SURABAYA', regex=False)
    .str.replace('KOTA BATU', 'BATU', regex=False)
)

df_final
```
